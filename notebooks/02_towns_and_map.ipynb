# Towns and planning areas

Median price per sqm over the latest 12 months, and the change against the same 12 months two years earlier. Groups with fewer than 50 sales in either window are left out.

```python
import sys
from pathlib import Path

# Make the repository importable when the kernel starts in notebooks/.
ROOT = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / "pyproject.toml").exists())
sys.path.insert(0, str(ROOT))
```

```python
from config.settings import get_settings
from pipeline.common.db import get_engine
from pipeline.eda import data, figures

figures.apply_style()
engine = get_engine()
FIGURES = get_settings().reports_dir / "figures"
```

```python
import pandas as pd

last = pd.Timestamp(data.last_month(engine))
WINDOW = f"{last - pd.DateOffset(months=11):%b %Y}–{last:%b %Y}"
WINDOW
```

## Town ranking

```python
towns = data.town_ranking(engine)
fig = figures.town_ranking(towns, WINDOW)
figures.save_figure(fig, "town_ranking", FIGURES)
```

```python
towns.round(3)
```

## Planning areas

```python
areas = data.planning_area_prices(engine)
fig = figures.planning_area_map(areas, WINDOW)
figures.save_figure(fig, "planning_area_map", FIGURES)
```

```python
areas.drop(columns="geom").dropna().sort_values("latest_median_psm", ascending=False).round(3)
```
