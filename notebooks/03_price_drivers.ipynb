# Lease, floor level and distance to the MRT

Each sale since 2015 is compared with the median price per sqm of its town, flat type and year. A ratio above 1 means the sale was dearer per sqm than similar flats sold nearby in the same year, which takes out most of the effect of location, flat type and market level.

```python
import sys
from pathlib import Path

# Make the repository importable when the kernel starts in notebooks/.
ROOT = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / "pyproject.toml").exists())
sys.path.insert(0, str(ROOT))
```

```python
from config.settings import get_settings
from pipeline.common.db import get_engine
from pipeline.eda import data, figures

figures.apply_style()
engine = get_engine()
FIGURES = get_settings().reports_dir / "figures"
```

```python
import pandas as pd

last = pd.Timestamp(data.last_month(engine))
WINDOW = f"{last - pd.DateOffset(months=11):%b %Y}–{last:%b %Y}"
WINDOW
```

## Lease decay

```python
lease = data.lease_decay(engine)
fig = figures.lease_decay(lease)
figures.save_figure(fig, "lease_decay", FIGURES)
```

## Floor-level premium

```python
by_storey, by_rel = data.floor_premium(engine)
fig = figures.floor_premium(by_storey, by_rel)
figures.save_figure(fig, "floor_premium", FIGURES)
```

```python
mrt = data.mrt_distance(engine)
fig = figures.mrt_distance(mrt, WINDOW)
figures.save_figure(fig, "mrt_distance", FIGURES)
```
